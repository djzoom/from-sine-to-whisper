## 2. How to read a paper efficiently — the three-pass method

### First pass (5-10 minutes): get your bearings
1. **Title + abstract**: what problem does it solve? What method does it propose?
2. **Figures and tables**: look only at the figures, not the text. Figure 1 is usually a diagram of the method; the last figure is usually a comparison of results.
3. **Conclusion**: the authors' own account of "what we achieved".
4. Decide: is it worth a close read?

### Second pass (1-2 hours): understand it
- Read the introduction (background and motivation)
- Read the method section closely: you should be able to derive every formula by hand and name every symbol
- Compare with the code (if it's open source): paper formula ↔ function name in the code
- Skip proofs you can't follow yet; mark them and come back

### Third pass (half a day to a day): reproduce it
- Derive the core formulas on paper
- Write the core algorithm from a blank file (Aurora style: no wrapper)
- Run a toy experiment to check that your output matches the paper

## 3. The structure of a research paper

The skeleton of a standard 8-page conference paper:

```
Abstract (~150 words)
  └─ problem / method / key result / conclusion (one sentence each)

1. Introduction (~1 page)
  └─ motivation → limitations of existing methods → our contributions

2. Related Work (~0.5 page)
  └─ compare your method with 3-5 related papers and make clear "how it differs"

3. Method (the core, ~2-3 pages)
  └─ problem definition (table of notation) → architecture/algorithm → key formulas

4. Experiments (~2 pages)
  └─ datasets / evaluation metrics → main results table → ablation study

5. Conclusion (~0.5 page)
  └─ restate the contributions, point out limitations and future directions

References
  └─ 8-30 papers, managed with BibTeX
```

**Technical report vs paper**: if your work isn't yet up to publication standard, writing a tech report first
and posting it to arXiv is a great way to build up a track record — Whisper itself was released as nothing more than an arXiv preprint.

## 4. Where to submit

### Major audio / speech AI conferences

| Conference | Full name | Deadline (typical) | Acceptance rate | Good fit for Aurora on |
|---|---|---|---|---|
| **ICASSP** | IEEE Int. Conf. on Acoustics, Speech and Signal Processing | September-October each year | ~50% | signal processing, feature extraction, ASR |
| **Interspeech** | Annual Conf. of ISCA | February-March each year | ~50% | speech recognition, synthesis, speaker work |
| **ICLR** | Int. Conf. on Learning Representations | September-October each year | ~30% | audio + deep learning crossover |
| **NeurIPS** | Neural Information Processing Systems | May-June each year | ~25% | generative models, foundation models |
| **ICML** | Int. Conf. on Machine Learning | January-February each year | ~28% | algorithmic theory + audio |

### arXiv — the preprint server
- No peer review: it's public as soon as you upload, usually visible after one working day
- Address: `arxiv.org`, categories `cs.SD` (Sound), `cs.LG`, `eess.AS`
- Strategy: post to arXiv first to stake your claim, and submit to a conference at the same time — the two don't conflict

### Review timeline (ICASSP as an example)
```
September → submission deadline
November  → reviews come back
December  → revisions / responses
January   → final notification
May       → the conference (in person, or online poster/oral)
```

## 5. Managing the literature: don't pile PDFs into folders

Once you've read a lot of papers, the biggest pain is "where did I see that result?".
Manage them with a tool from day one, and citing them when you write later will save you a great deal of time.

### Recommended tools

| Tool | Strengths | Best for |
|---|---|---|
| **Zotero** (free) | one-click save from a browser extension, automatic metadata, BibTeX export | your main library — essential for writing papers |
| **Obsidian + PDF** | notes linked to PDFs, bidirectional links, local storage | in-depth reading notes |
| **Semantic Scholar** | AI-assisted related-paper recommendations, citation-network visualization | discovering related papers |

## 6. Peer review: the core mechanism of publishing

### The review process (double-blind review as an example)

```
Submission
  ↓ 1-2 weeks
Area Chair assigns it to 3 reviewers
  ↓ 4-6 weeks
3 reviews (each usually 1-3 pages)
  ↓ 1-2 weeks
Authors write a rebuttal (a response, usually ≤1 page)
  ↓ 2 weeks
Reviewers may revise their scores → AC makes a recommendation
  ↓ 1 week
Program Committee makes the final decision
```

### Three kinds of review comment

| Type | Signal | How to handle it |
|---|---|---|
| Technical error | a wrong derivation, a flawed experimental design | must be fixed — thank them for pointing it out |
| Missing experiment | "no comparison with method X" | if you can run it, run it and add it; otherwise explain why |
| Subjective opinion | "I don't think this problem matters" | politely explain the motivation, and cite work that shows it matters |

### Principles for writing a rebuttal

1. **Thank first, then respond** — even when the reviewer is wrong
2. **Respond to each comment separately**; don't lump them together
3. **Back it up with numbers** — "we ran the extra experiment and WER improved by 0.8%" is more convincing than "we believe..."
4. **Stay calm** — reviewers are volunteers, not enemies

## 8. Practical routes to collaborating with academic institutions

### Route 1: open-source reproduction
- Reproduce a lab's paper, publish the code on GitHub, and write a detailed README
- The authors usually notice, and a good reproduction can open the door to getting in touch
- Aurora itself is a high-quality reproduction project

### Route 2: contacting a professor / researcher
The key principle when writing the email: **specific > enthusiastic**

```
Subject: Re: [paper title] — code reproduction + one question

Dear Dr. [name],

I've recently been reproducing [specific method] from your paper [paper title],
and ran into [a specific problem: e.g. the boundary condition of Equation (5) in Section 3.2].
My implementation is here: [GitHub link]

Also, if your group runs any community-facing open experiments (such as data labeling or benchmark testing),
I'd be glad to contribute.

[name]
```

A specific question + existing work → a far higher reply rate than "I'd like to join your lab".

### Route 3: joining open dataset / benchmark projects
- **LibriSpeech** (OpenSLR): an open ASR dataset; you can submit WER results
- **SUPERB**: a benchmark of speech tasks; submit your system's scores
- **HEAR** (Holistic Evaluation of Audio Representations): an evaluation for audio embeddings

### Route 4: applying as a visiting student / remote collaborator
- Some universities offer a "visiting student" status, which lets you co-author papers across institutions
- What it takes: solid coding skills + a concrete idea + a supervisor with room to take you on

## 9. How to write Related Work

Related Work is the easiest section to get wrong — many people turn it into a "list of methods".
A good Related Work section makes clear **how your work relates to theirs**, not just what they did.

### Three ways to organize Related Work

1. **Group by technical approach** (not in chronological order)
   Split the methods into 2-3 schools; for each group say what they share and where they fall short, then say how you address that

2. **One sentence of difference per paper**
   "Samson et al. (2022) solved problem Y with method X, but their accuracy drops by N% in scenario Z.
   This work addresses that limitation through Q."

3. **End with one sentence of positioning**
   "Unlike the methods above, this work is the first to satisfy both constraint A and constraint B."

### What Related Work should cover for Aurora

If you write a report on Aurora's audio feature extraction, Related Work should mention at least:
- the classic MFCC approach (Davis & Mermelstein 1980)
- end-to-end feature learning (wav2vec 2.0)
- where your method stands: interpretable, derivable, no-dependency

## 10. Keeping up with the field: 30 minutes a week is enough

You don't need to read every paper — you just need to **know what's happening**.

### Recommended feeds

| Channel | How often | What for |
|---|---|---|
| `arxiv.org/list/eess.AS/recent` | once a week | the latest audio signal processing preprints |
| `paperswithcode.com/sota/speech-recognition` | once a month | changes in the SOTA leaderboard |
| Hugging Face Blog | once a week | new model releases |
| ACL Anthology / IEEE Xplore | after each conference | batch-download the papers you're interested in |
| Google Scholar citation alerts | automatic | track who is citing the papers you follow |

### A habit you can actually keep

Every Monday morning, 30 minutes:
1. Open arXiv eess.AS, skim the titles, and add the interesting ones to Zotero
2. Pick 1 paper for a first-pass read (5-10 minutes, to decide whether it deserves a close read)
3. Write its title and a one-sentence summary into `docs/paper_notes/weekly_log.md`

After a year you'll have 50 organized entries — and that in itself is a piece of work you can show someone (in an interview, to a colleague, or to your future self).

## 12. At the conference: getting the most out of attending

If you get the chance to attend ICASSP or Interspeech, whether online or in person,
preparing ahead can double what you get out of it:

**Before (1 week)**
- Download the proceedings and add the papers you're interested in to Zotero
- Mark 3-5 papers whose authors you especially want to meet
- Prepare a one-sentence introduction: "I'm at X, working on Y, and lately I've been doing Z"

**During**
- Poster sessions are the best chance to meet authors: just walk up and say "I read your paper and I have a question..."
- Note down business cards or LinkedIn profiles, and follow up with a short email that same evening

**After (within 1 week)**
- Write a 500-word conference report (which work surprised you most, and why)
- Sort the business cards into your contacts, and add the authors you're interested in to your Google Scholar citation alerts
