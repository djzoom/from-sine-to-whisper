## 2c. Open research problems in audio AI

If you pick direction A or are looking for a way into a paper, these are directions that still have room to explore as of 2026:

| Problem | State of play | Possible way in |
|---|---|---|
| ASR for low-resource languages | data is scarce, WER is still high | data augmentation + cross-lingual transfer |
| Music understanding | harmony and rhythm analysis accuracy is low | improve chroma / beat features |
| Real-time TTS voice cloning | latency > 500ms, quality vs speed tradeoff | streaming vocoders |
| Audio event detection | accuracy drops when several events overlap | multi-label classification + temporal localization |
| ASR robustness | WER rises sharply in noisy environments | data augmentation (extending SpecAugment) |
| Evaluating audio foundation models | no unified benchmark | a GLUE-style audio evaluation suite |

Every one of these directions has sub-problems with a low barrier to entry — a good thesis / short conference paper
often only needs to **improve one metric by 1-3% on one specific dataset**, with a clear ablation.

## 3c. Building a public profile (no need to become an influencer)

A public profile isn't marketing; it's how the right people find you.
Each of the three things below takes only 1-2 hours a month, but keep at it over the long run and the returns compound like interest:

### 1. Keep your GitHub green

At least 3 meaningful commits a week.
What counts as "meaningful": the commit message records a specific technical decision,
not `fix stuff` or `update readme`.

### 2. One short post a month

It doesn't need to be a full tutorial — 500 words is plenty:
- "This week I finally understood the STFT window-function tradeoff"
- "Debugging notes on the CTC forward algorithm"
- "Whisper fine-tuning pitfalls: how batch size affects WER"

The platform doesn't matter; what matters is **consistency**. 12 posts > 1 perfect long-read.

### 3. Leave a trail in the community

- Open an issue or PR on the open-source implementation of a paper you've read (even a documentation fix)
- Leave a substantive comment on an audio model page on Hugging Face
- Answer an audio-AI question on Stack Overflow / GitHub Discussions

## 3d. How to know you're ready

Don't wait until you're "completely ready" to interview or submit a paper — that day never comes.
Here are "good enough" criteria along two dimensions:

### Interviews
- [ ] Can write a naive DFT from a blank file in 20 minutes
- [ ] Can derive the attention formula in 15 minutes
- [ ] Can describe the Aurora project in 5 minutes (with numbers, and with limitations)
- [ ] Have at least 1 demo you can show (one the interviewer can run)
- [ ] At least 10 items in L98's `deliverables` are ✅

### Paper submission
- [ ] Can clearly distinguish your contribution from 3 related papers
- [ ] Have a reproducible experiment (same code and data, same result)
- [ ] Have a number that beats the baseline (even by just 1%, with an ablation)
- [ ] Your abstract lets someone unfamiliar with your work understand what you did in 2 minutes

## 4. A system of habits for continued growth

After the course, these are the three most effective habits for keeping your growth going:

**① One reading note a week**
Choose the paper most relevant to your current work, read it with the three-pass method from L95,
and save a 200-word note in `docs/paper_notes/`.
A year from now you'll have 50 organized reading notes — a portfolio in its own right.

**② One "skills test" a month**
Pick a random Aurora module and rewrite its core algorithm from a blank file,
without looking at your old code. Time yourself and note where you got stuck.
This habit keeps your knowledge from decaying, and it's the best warm-up before an interview.

**③ Share your work in public**
Every time you finish a new feature, write a short blog post (500 words) or record a demo video.
Sharing in public pays off three ways:
- The Feynman technique: you only find out what you haven't understood when you write it down
- Visibility: the technical community discovers you
- A record: six months on, you can see the path of your own growth
