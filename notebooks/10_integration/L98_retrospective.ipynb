## 6. Learning velocity: estimated vs actual

How did each month's progress compare with what you expected? An honest record helps you estimate your next project more accurately.

```python
learning_velocity = {
    # Format: (estimated days, actual days, the single hardest concept)
    "Month 1 · Audio Core":       (30, None, None),
    "Month 2 · Deep Learning":    (30, None, None),
    "Month 3 · ASR":              (30, None, None),
    "Month 4 · Music":            (30, None, None),
    "Month 5 · LLM/RAG":         (30, None, None),
    "Month 6 · Integration":      (30, None, None),
}

print("Month                  Est.   Actual  Hardest concept")
print("-" * 60)
for phase, (est, actual, hard) in learning_velocity.items():
    actual_str = str(actual) if actual is not None else "?"
    hard_str   = hard if hard is not None else "not filled in"
    ratio = f"({actual/est*100:.0f}%)" if actual is not None else ""
    print(f"{phase[:22]:<22} {est:>4}d  {actual_str:>4}d {ratio:>6}  {hard_str}")
```

```python
# Learning-curve plot: estimated vs actual days (delivers on the "measuring the learning curve" promise in the subtitle)
# Guard: only plots automatically once every actual field is filled in
if all(v[1] is not None for v in learning_velocity.values()):
    try:
        import matplotlib.pyplot as plt
        import matplotlib

        phases = list(learning_velocity.keys())
        labels = [p.split('·')[1].strip() if '·' in p else p for p in phases]
        estimated = [v[0] for v in learning_velocity.values()]
        actual    = [v[1] for v in learning_velocity.values()]

        x = list(range(len(phases)))
        fig, ax = plt.subplots(figsize=(9, 4))
        width = 0.35
        bars_est = ax.bar([xi - width/2 for xi in x], estimated, width,
                          label='Estimated days', color='#4C72B0', alpha=0.85)
        bars_act = ax.bar([xi + width/2 for xi in x], actual,    width,
                          label='Actual days', color='#DD8452', alpha=0.85)

        ax.set_xticks(x)
        ax.set_xticklabels(labels, rotation=20, ha='right')
        ax.set_ylabel('Days')
        ax.set_title('Aurora learning velocity over 6 months: estimated vs actual (your learning curve in numbers)')
        ax.legend()
        ax.axhline(30, color='gray', linestyle='--', linewidth=0.8, alpha=0.5)

        for bar in bars_est:
            ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3,
                    str(int(bar.get_height())), ha='center', va='bottom', fontsize=8)
        for bar in bars_act:
            ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.3,
                    str(int(bar.get_height())), ha='center', va='bottom', fontsize=8)

        plt.tight_layout()
        plt.show()
    except ImportError:
        print("matplotlib is not installed, skipping the plot. Run pip install matplotlib and try again.")
else:
    print("📊 Plot not active yet: first fill in every actual (actual days) field in learning_velocity")
    for phase, (est, actual, hard) in learning_velocity.items():
        if actual is None:
            print(f"  ⬜ {phase}: actual days not filled in")
```
